```python
file = open("sales_data.csv", "r")
lines = file.readlines()

#print(lines) --- IGNORE ---
print(lines[:5])  # Print the first 5 lines of the file
```

```python
newData = []
for line in lines[1: ]:
    field = line.strip().split(",")
    field[5] = int(field[5])
    field[6] = int(field[6])
    newData.append(field)

#print(newData)
print(newData[:5])  # Print the first 5 records of the processed data
```

```python
totalSales = 0
for row in newData:
    totalSales += row[5] * row[6]
print("Total Revenue: $", totalSales)
```

```python
total_orders = len(newData)

print("Total Orders: ", total_orders)
```

```python
total_quantity = 0
for row in newData:
    total_quantity += row[5]
    
print("Total Quantity Sold: ", total_quantity)
```

```python
category_revenue = {}

for row in newData:
    category = row[4]
    revenue = row[5] * row[6]

    if category in category_revenue:
        category_revenue[category] += revenue
    else:
        category_revenue[category] = revenue

print("Revenue by Category:")
for cat, rev in category_revenue.items():
    print(f"{cat}: $ {rev}")
```

```python
product_revenue = {}

for row in newData:
    product = row[3]
    revenue = row[5] * row[6]

    if product in product_revenue:
        product_revenue[product] += revenue
    else:
        product_revenue[product] = revenue

print("Revenue by Product:")
for pro, rev in product_revenue.items():
    print(f"{pro}: $ {rev}")
```

```python
product_quantity = {}

for row in newData:
    product = row[3]
    quantity = row[5]

    if product in product_quantity:
        product_quantity[product] += quantity
    else:
        product_quantity[product] = quantity

highest_sold = None
max_quantity = 0

for product, qty in product_quantity.items():
    if qty > max_quantity:
        max_quantity = qty
        highest_sold = product
print("Highest Sold Product: ", highest_sold, "\nQuantity Sold: ", max_quantity)
```

```python
customer_revenue = {}

for row in newData:
    customer = row[2]
    revenue = row[5] * row[6]

    if customer in customer_revenue:
        customer_revenue[customer] += revenue
    else:
        customer_revenue[customer] = revenue

top_customer = None
max_revenue = 0

for cus, rev in customer_revenue.items():
    if rev > max_revenue:
        max_revenue = rev
        top_customer = cus

print("Top Customer: ", top_customer, "\nRevenue: $", max_revenue)
```

```python
daily_revenue = {}

for row in newData:
    date = row[1]
    revenue = row[5] * row[6]

    if date in daily_revenue:
        daily_revenue[date] += revenue
    else:
        daily_revenue[date] = revenue

print("Daily Revenue: ")
for date, rev in daily_revenue.items():
    print(f"{date}: $ {rev}")
```

```python
from datetime import datetime

start = input("Enter start date (YYYY-MM-DD): ")
end = input("Enter end date (YYYY-MM-DD): ")

start_date = datetime.strptime(start, "%Y-%m-%d")
end_date = datetime.strptime(end, "%Y-%m-%d")

if start_date > end_date:
    raise ValueError("Start date must be before or equal to the end date.")

total_date_revenue = 0

for row in newData:
    order_date = datetime.strptime(row[1], "%m/%d/%Y")

    if start_date <= order_date <= end_date:
        total_date_revenue += row[5] * row[6]

print("Revenue from", start, "to", end, "is $", total_date_revenue)
```

```python
elecFile = open("electronics_sales.csv", "w")

elecFile.write("OrderID,OrderDate,CustomerName,ProductName,Category,Quantity,Price\n")

for row in newData:
    if row[4] == "Electronics":
        row_str = ",".join([str(item) for item in row])
        elecFile.write(row_str + "\n")

elecFile.close()

print("All electrical items saved to electronics_sales.csv")
```
